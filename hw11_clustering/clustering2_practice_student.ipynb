## Краткая теория

### Типы точек в DBSCAN

DBSCAN задаётся двумя параметрами: радиусом окрестности `eps` (ε) и минимальным числом точек `min_samples`.

- **Корневая точка** (core): в её ε-окрестности ≥ `min_samples` точек.
- **Граничная точка** (border): в её ε-окрестности < `min_samples`, но она достижима из корневой.
- **Шумовая точка** (noise): не является ни корневой, ни граничной → метка **-1**.

### Как работает алгоритм

1. Для каждой непосещённой точки найти её ε-окрестность.
2. Если точек ≥ `min_samples` — начать новый кластер.
3. Рекурсивно расширить кластер через цепочки плотно достижимых точек.
4. Оставшиеся точки → шум (метка -1).

### Как выбрать eps: k-distance plot

Для каждой точки вычисляют расстояние до её k-го ближайшего соседа (k = `min_samples`). Отсортированный по убыванию график этих расстояний называется **k-distance plot**. Резкий перегиб («локоть») указывает на оптимальное значение `eps`.

**1. Когда DBSCAN выигрывает у K-Means?**

> DBSCAN имеет преимущество перед K-Means когда кластеры имеет сложную структуру(например:полумесяц или другие извилистые структуры).Также DBSCAN хорошо справляется с большим шумом и выбросами(помечая изолированные точки как щум и исключая их из кластера).Ну и DBSCAN определяет количество кластеров автоматически на основе плотности данных.

**2. Когда K-Means предпочтительнее?**

> K-Means предпочтительнее при разной плотности кластеров в датасете. Также K-means предпочтительнее на данных высокой размерности(как в датасете Wine имея 13 признаков)настроить один глобальный радиус eps для DBSCAN становится крайне сложно.K-Means работает быстрее и требует меньше памяти на огромных датасетах, в то время как классическому DBSCAN нужно просчитывать попарные расстояния или строить пространственные деревья.

**3. Почему масштабирование обязательно для DBSCAN?**

> Без масштабирования невозможно подобрать адекватное значение eps, которое бы учитывало вклад всех признаков поровну.

**4. Как интерпретировать шумовые точки в датасете Wine?**

> Эксклюзивные, уникальные вина, чьи химические характеристики сильно отличаются от стандартных профилей основных сортов.

**5. Можно ли корректно сравнивать силуэт DBSCAN и K-Means напрямую? Почему?**

> Нет, т.к. метрика силуэта изначально заточена под круглые (сферические) кластеры. Если DBSCAN идеально найдет сложный нелинейный кластер (например: луну), силуэт у него всё равно будет низким, хотя кластеризация идеальна.
